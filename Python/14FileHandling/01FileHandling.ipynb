# File Read & Write

```python
readFile = open('test.txt', 'r')

print(readFile.read())

readFile.close()

writeFile = open('test.txt', 'w')

writeFile.write('This is File Handling Write method!...')

writeFile.close()

readFile = open('test.txt', 'r')

print(readFile.read())

print(readFile.read())
```

# File Read & Append

```python
readFile = open('test.txt', 'r')

print(readFile.read())

readFile.close()

writeFile = open('test.txt', 'a')

writeFile.write(' This is File Handling Write method!...')

writeFile.close()

readFile = open('test.txt', 'r')

print(readFile.read())

print(readFile.read())
```

# File Delete

```python
import os

file = 'sample.txt'

if os.path.exists(file):
    print(file , ' - File is There')

    os.remove(file)

    print(file, ' - removed Successfully!..')

else:
    print(file, ' - File is not in your Folder!...')
```
