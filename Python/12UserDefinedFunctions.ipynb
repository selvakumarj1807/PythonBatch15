# User Defined Function

# 01 No Return Type With Out Argument

```python
# Create the Function

def welcome():
    print('Welome Python Programming Language!...')
```

```python
# Call the Function

welcome()

print('------------------------------')

welcome()
```

```python
# Create the Function
# a + b = a + b --> 10 + 2 = 12 ==> a, '+', b, '=', a + b

def addTwoNumber():
    a = int(input('Enter a Value : '))

    b = int(input('Enter a Value : '))

    print('')

    print(f"{a} + {b} = {a + b}") 
```

```python
# Call the Function

addTwoNumber()

print('--------------------------')

addTwoNumber()
```

# 02 No Return Type With Argument

```python
# Create the Function

def addTwoNumber(a, b):
    print(f"{a} + {b} = {a + b}") 
```

```python
# Call the Function

addTwoNumber(10, 2)

print('--------------------------')

addTwoNumber(20, 2)
```

# 03 Return Type With Out Argument

```python
# Create the Function
# a + b = a + b --> 10 + 2 = 12 ==> a, '+', b, '=', a + b

def addTwoNumber():
    a = int(input('Enter a Value : '))

    b = int(input('Enter a Value : '))

    c = a + b

    return c
```

```python
# Call the Function

print('Call the Function Using Print Statement : ', addTwoNumber())

print('--------------------------')

result01 = addTwoNumber()

print('Call the Function Using Variable : ', result01)
```

# 04 Return Type With Argument

```python
# Create the Function

def additionOfTwoNumber(a, b):
    c = a + b

    return c
```

```python
# Call the Function

print('Call the Function Using Print Statement : ', additionOfTwoNumber(10, 2))

print('--------------------------')

result01 = additionOfTwoNumber(20, 2)

print('Call the Function Using Variable : ', result01)

print('--------------------------')

additionOfTwoNumber(30, 2)
```

# Arbitrary Argument Function

```python
def userData(*data):
    print(data, type(data))
```

```python
userData('Selva', 27, 'male')
```

# Pass List Values

```python
def userData(data):
    print(data, type(data))
```

```python
userData(['Selva', 27, 'male'])
```

# Key Word Argument Function

```python
def userData(name, age, gender):
    print(name, 'your age is', age, 'and your gender is', gender)
```

```python
userData('Selva', 27, 'male')

print('----------------------------------')

userData(name='Selva', gender='male', age=27)
```

# Key Word Arbitrary Argument Function

```python
def userData(**data):
    print(data, type(data))
```

```python
userData(name='Selva', gender='male', age=27)
```

# Default Parameter Function

```python
def userData(name, age, gender, state='TN'):
    print(name, 'your age is', age, 'and your gender is', gender, '-', state)
```

```python
userData('Selva', 27, 'male', 'KL')

print('----------------------------------')

userData('Selva', 27, 'male')
```

# Lambda Function

```python
addTwoNumber = lambda a, b: a + b

print(addTwoNumber(10, 2))
```

```python
addTwoNumber = lambda a, b: f"{a} + {b} = {a + b}"

print(addTwoNumber(10, 2))

print(addTwoNumber(20, 2))

print(addTwoNumber(30, 2))
```
