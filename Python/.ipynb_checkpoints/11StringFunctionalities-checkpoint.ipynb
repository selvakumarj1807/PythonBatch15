```python
college = "St Josephs's College"

print(college)

paragraph = '''
Based on 2025–26 Gross State Domestic Product (GSDP) estimates, 
the top three richest states in India are Maharashtra (approx. ₹45.3 lakh crore), 
Tamil Nadu (approx. ₹31.1 lakh crore), and Uttar Pradesh (approx. ₹29.7 lakh crore). 
These states are top contributors to India's GDP, driven by strong manufacturing, 
IT, and financial services sectors. 
'''

print(paragraph)
```

# Sting Manipulation

```python
message = 'Welcome to Python'

print(message)

print(message[1])

print(message[11:])

print(message[8:10])
```

# String Functions

```python
message = 'Welcome to Python'

# length

print(len(message))

# Count

print(message.count('o'))

# Find

print(message.find('o'))
```

```python
message = 'wELcoME tO pYTHoN sESsIoN'

print(message)

# Upper

print(message.upper())

# Lower

print(message.lower())

# Title

print(message.title())

# Capitalize

print(message.capitalize())
```

```python
value01 = 'PYTHON'

print(value01.isupper())

value02 = 'python'

print(value02.islower())

value03 = 'Python'

print(value03.isalpha())

value04 = 'PythonBatch15'

print(value04.isalnum())
```

```python
# strip

stripvalue01 = 'XX95X00XX'

print(stripvalue01.strip('X'))

# lstrip

stripvalue01 = 'XX95X00XX'

print(stripvalue01.lstrip('X'))

# rstrip

stripvalue01 = 'XX95X00XX'

print(stripvalue01.rstrip('X'))
```

```python
# spit

v1 = 'Ram, Sam, Dhara'

splitValue01 = v1.split(', ')

print(splitValue01)
```

```python
# Join

v1 = ['Ram', 'Sam', 'Dhara']

joinValue01 = ' / '.join(v1)

print(joinValue01)
```
