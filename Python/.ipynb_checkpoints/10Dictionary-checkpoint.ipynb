# Dictionary
------------

1. Dictionary is a collection of Data Type.
2. declare - {'key': 'value'}
3. Dictionary are used to store many values & Different Data Type Values in a single Variable. Based On Non Index.
4. Dictionary are Mutable(Changable)
5. Dictionary are Used for Remove Douplicate Values

```python
userData = {'name': 'Ram', 'age': 27, 'mobile': 9500912258}

print(userData, type(userData))
```

```python
print(userData['name'])
```

```python
for i in userData:
    print(i, ' - ', userData[i])
```

# Value change after variable declaration

```python
userData = {'name': 'Ram', 'age': 27, 'mobile': 9500912258}

print(userData)

print('--------------------')

print(userData['name'])

userData['name'] = 'Dharani'

print(userData)
```

# Add Values After Variable Declaration

```python
userData = {'name': 'Ram', 'age': 27, 'mobile': 9500912258}

print(userData)

print('--------------------')

userData.update({'gender': 'male', 'city': 'Chennai'})

print(userData)
```

# Remove Values

```python
userData = {'name': 'Ram', 'age': 27, 'mobile': 9500912258}

print(userData)

print('--------------------')

userData.pop('mobile')

print(userData)
```

```python
users = {
    'user01': {'userId': 101, 'name': 'Ram'}, 
    'user02': {'userId': 102, 'name': 'Sam'},
    'user03': {'userId': 103, 'name': 'Sara'}
}
```

```python
print(users)
```

```python
print(users['user01'])
```

```python
print(users['user01']['userId'])
print(users['user01']['name'])
```

```python
print(users['user02']['userId'])
print(users['user02']['name'])
```

```python
indx = 1

for i in users:
    print(indx, ':', i, users[i]['userId'])
    print(indx, ':', i, users[i]['name'])

    indx += 1
    print('')
```
