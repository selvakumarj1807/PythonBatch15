# Inheritance

# Single Inheritance

```python
class Mother:
    def play(self):
        print('Mother have Playing Skill!...')

    def cook(self):
        print('Mother have Cooking Skill!...')

class Son(Mother):
    def read(self):
        print('Son have Reading Skill!...')
```

```python
object01 = Son()

object01.read()

object01.play()

object01.cook()
```

# Multiple Inheritance

```python
class Father:
    def play(self):
        print('Father have Playing Skill!...')

    def write(self):
        print('Father have Writing Skill!...')

class Mother:
    def play(self):
        print('Mother have Playing Skill!...')

    def cook(self):
        print('Mother have Cooking Skill!...')

class Son(Mother, Father):
    def read(self):
        print('Son have Reading Skill!...')
```

```python
object01 = Son()

object01.read()

object01.play()

object01.cook()

object01.write()

object01.play()
```

# Multi Level Inheritance

```python
class Father:
    def play(self):
        print('Father have Playing Skill!...')

    def write(self):
        print('Father have Writing Skill!...')

class Mother(Father):
    def play(self):
        print('Mother have Playing Skill!...')

    def cook(self):
        print('Mother have Cooking Skill!...')

class Son(Mother):
    def read(self):
        print('Son have Reading Skill!...')
```

```python
object01 = Son()

object01.read()

object01.play()

object01.cook()

object01.write()

object01.play()
```
