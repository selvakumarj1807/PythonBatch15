# Create the Class

```python
var = 10

class Student:
    name = 'Student Name'
    age = 'Student Age'

    def details(self):
        print('Hi', self.name, 'your age is', self.age)
```

# Create the Object & Call the Function Using Object

```python
student01 = Student()

print('Before Value Assign : ')

student01.details()

student01.name = 'Sara'

student01.age = 25

print('After Value Assign : ')

student01.details()
```

# Creat the Clas

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def details(self):
        print('Hi', self.name, 'your age is', self.age)
```

```python
student01 = Student('Sara', 25)

student01.details()
```

```python
student02 = Student('Lara', 22)

student02.details()
```
