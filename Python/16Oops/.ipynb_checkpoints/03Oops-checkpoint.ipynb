# Abstraction & Encapsulation

```python
class Library:
    def __init__(self, books):
        self.books = books

    def displayBook(self):
        print('Our Display Book')
        
        for i in self.books:
            print(i, end=', ')

    def borrowBook(self, book):
        if book in self.books:
            self.books.remove(book)
            print(f"Get your {book} Now!...")
        else:
            print(f"{book} not in your Library!...")

    def reciveBook(self, book):
        self.books.append(book)
        print(f"{book} Recived Successfully!....")
        
```

```python
print('Limbrary Management')

msg = '''
1. Display Book
2. Borrow Book
3. Recive Book
'''

books = ['C', 'C++', 'Java']

library01 = Library(books)

while True:
    print(msg)

    choice = input('Enter your Choice (1/2/3): ')
    
    if choice == '1':
        library01.displayBook()
    
    elif choice == '2':
        book = input('Enter your Borrow Book Name : ')
        library01.borrowBook(book)
    
    elif choice == '3':
        book = input('Enter your Recive Book Name : ')

        library01.reciveBook(book)
    
    else:
        print('Enter a valid Choice')
        print('Thank you Come Again!...')
        break
```
