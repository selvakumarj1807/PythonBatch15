# Date Time Module

```python
import datetime as dt
```

```python
currentDate = dt.date.today()

print(currentDate)
```

```python
userDate = dt.date(2026, 3, 21)

print(userDate)
```

```python
currentDateTime = dt.datetime.now()

print(currentDateTime)
```

```python
userDateTime = dt.datetime(2025, 8, 21, 12, 59)

print(userDateTime)
```

```python
currentDateTime = dt.datetime.now()

print(currentDateTime)

print('------------------------------')

currentDate = currentDateTime.date()

print(currentDate)

currectTime = currentDateTime.time()

print(currectTime)
```

# To find the number of days for given date

```python
currentDate = dt.date.today()

userDate = dt.date(2026, 3, 21)

diffTwoDate = currentDate - userDate

print(diffTwoDate)
```

# DateTime Format

```python
currentDateTime = dt.datetime.now()

print(currentDateTime)

print('--------------------------')

format01 = currentDateTime.strftime("%B")

print(format01)

format02 = currentDateTime.strftime("%d / %b / %Y - %B")

print(format02)
```

# Random Module

```python
import random as rn
```

```python
r1 = rn.randint(1000, 9999)

print(r1)

print('PR'+str(r1))
```

```python
team = ['team 01', 'team 02', 'team 03']

r2 = rn.choice(team)

print(r2)
```
