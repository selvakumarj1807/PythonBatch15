# Python Exception Handling

# try & except

```python
try:
    a = 10

    b = 0
    
    print(a / b)

except:
    print('Divsion Cannot be Zero')
```

# To Find the Exception

```python
try:
    a = 10

    b = 0
    
    print(a / b)

except Exception as e:
    print('Program have Some Issues : ', e)
```

# try & except & else

```python
try:
    a = 10

    b = 5
    
    c = a / b

except Exception as e:
    print('Program have Some Issues : ', e)
    
else:
    print(c)
```

# try & except & else & finally

```python
try:
    a = 10

    b = 5
    
    c = a / b

except Exception as e:
    print('Program have Some Issues : ', e)
    
else:
    print(c)

finally:
    print('Thank You Come Again!...')
```
